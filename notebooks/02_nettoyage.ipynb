```python
import pandas as pd
import sys

sys.path.insert(0, "..")
from src.cleaning import normalize_survey_columns, parse_years_experience

df = pd.read_csv("../data/raw/survey_results_public.csv")
df = normalize_survey_columns(df)
colonnes_utiles = [
    "YearsCodePro",
    "Country",
    "EdLevel",
    "LanguageHaveWorkedWith",
    "Employment",
    "ConvertedCompYearly",
]
df = df[colonnes_utiles].copy()
```

```python
df = df.dropna(subset=["ConvertedCompYearly"])
```

```python
df = df[(df["ConvertedCompYearly"] >= 5000) & (df["ConvertedCompYearly"] <= 500000)]
```

```python
df["YearsCodePro"] = df["YearsCodePro"].apply(parse_years_experience)
df = df.dropna(subset=["YearsCodePro"])
```

```python
employment_normalise = (
    df["Employment"]
    .astype("string")
    .str.casefold()
    .str.replace(r"[^a-z]+", " ", regex=True)
    .str.strip()
)
df = df[employment_normalise.eq("employed full time")]
```

```python
df = df.dropna(subset=["LanguageHaveWorkedWith", "EdLevel", "Country"])
```

```python
df.to_csv("../data/processed/donnees_salaires_developpeurs.csv", index=False)
print(df.shape)
```
